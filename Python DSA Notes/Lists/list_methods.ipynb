## What are lists?

lists are basically:
- ordered
- mutable
- structured and linear data structures in python 
- also they are heterogeneous 
- and can also contain duplicates

## Commonly used list methods

```python
# Example use case of all commonly used methods

# Initial lists
lst = [1, 2, 3, 4, 5]
y = [-1, -2]
any_data = (7, 8, 9)  # tuple for demonstration

# append
lst.append(6)  # [1, 2, 3, 4, 5, 6]

# insert
lst.insert(0, 0)  # [0, 1, 2, 3, 4, 5, 6]

# remove (first occurrence)
lst.remove(1)  # [0, 2, 3, 4, 5, 6]

# pop by index
lst.pop(2)  # removes 3, [0, 2, 4, 5, 6]
lst.pop(-1)  # removes 6, [0, 2, 4, 5]
lst.pop()    # removes 5, [0, 2, 4]

# access by index
print(lst[1])  # 2

# extend
lst.extend(y)  # [0, 2, 4, -1, -2]

# index
print(lst.index(0))  # 0

# count
print(lst.count(0))  # 1

# reverse
lst.reverse()  # [-2, -1, 4, 2, 0]

# sort
lst.sort()  # [-2, -1, 0, 2, 4]

# sorted (creates new list)
z = sorted(lst, reverse=True)  # [4, 2, 0, -1, -2]

# copy
lst_copy = lst.copy()

# slicing
print(lst[2:5])    # [0, 2, 4]
print(lst[::2])    # [-2, 0, 4]
print(lst[::-1])   # [4, 2, 0, -1, -2]
print(lst[4:0:-1]) # [4, 2, 0, -1]

# length
print(len(lst))  # 5

# multidimensional list access
matrix = [[1, 2], [3, 4]]
print(matrix[0][1])  # 2

# min, max, sum
print(min(lst))  # -2
print(max(lst))  # 4
print(sum(lst))  # 3

# membership test
print(2 in lst)      # True
print(10 not in lst) # True

# create new list from any iterable
new = list(any_data)  # [7, 8, 9]

# clear (empties the list)
lst.clear()  # []

# delete by index
del lst_copy[0]  # removes first element

# delete by slice
del lst_copy[0:2]  # removes first two elements

# Show final states
print("lst:", lst)
print("lst_copy:", lst_copy)
print("z (sorted desc):", z)
print("new:", new)
print("matrix:", matrix)
```

## How Python Stores List Elements?

In Python, a list doesn’t store actual values directly. Instead, it stores `references (pointers)` to objects in memory. 

This means numbers, strings and booleans are separate objects in memory and the list just keeps their addresses.

That’s why modifying a mutable element (like another list or dictionary) can change the original object, while immutables remain unaffected.

## When built-in lists usually suffice:
`Python lists` dynamically resize and can hold mixed data types.

They provide excellent flexibility for most algorithmic problems.

## When you might use the array module:
You need a `memory-efficient array` storing only one data type (e.g., all integers).

For performance in large numeric datasets where memory overhead matters.

In some very low-level or performance-critical scenarios.

## Other array-like collections sometimes used:
`collections.deque` for double-ended queue operations.

`numpy arrays`, commonly in scientific computing or when you need fast vectorized array operations (less common in pure DSA interviews unless specified).
