### Why use Stack?
- A stack uses the Last In, First Out (LIFO) principle, which is also equivalent to First In, Last Out (FILO).
- Possible to do Undo/Redo Operations
- Push and Pop in O(1) time
- Enforced linear data order
- Effective memory management

### Why not to use Stack?

- Limited access
- Potential for overflow
- Not suitable for random access
- Limited capacity

### Basic operations of Stack:



```python
# Implementing Stack using Arrays

class Stack:
    def __init__(self, size):
        self.size = size
        self.top = -1
        self.stack = []

    def push(self, item):
        if len(self.stack) == self.size:
            print('Stack overflow')
        else:
            self.stack.append(item)
            self.top += 1
            print(f"{item} pushed to stack")

    def pop(self):
        if self.top == -1:
            print('Stack underflow')
            return None
        else:
            popped = self.stack.pop()
            self.top -= 1
            print(f"{popped} popped from stack")
            return popped

    def peek(self):
        if self.top == -1:
            print('Stack is empty')
            return None
        else:
            return self.stack[self.top]

    def display(self):
        if self.top == -1:
            print("Stack is empty")
        else:
            print("Stack (top to bottom):", self.stack[::-1])

# Interactive menu-driven stack operations
print("Please enter the desired size for the stack:")
size = int(input())
stack = Stack(size)
while True:
    print("\nChoose operation:\n1. Push\n2. Pop\n3. Peek\n4. Display\n5. Exit")
    choice = input("Enter choice: ")
    if choice == "1":
        item = int(input("Enter the element to push: "))
        stack.push(item)
    elif choice == "2":
        stack.pop()
    elif choice == "3":
        top_elem = stack.peek()
        if top_elem is not None:
            print(f"Top element: {top_elem}")
    elif choice == "4":
        stack.display()
    elif choice == "5":
        print("Exiting...")
        break
    else:
        print("Invalid choice, try again.")
```

```python
# Implementing stack using a Linked list:

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class StackLL:
    def __init__(self):
        self.top = None

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        print(f"{data} pushed to stack")

    def pop(self):
        if self.top is None:
            print("Stack underflow")
            return None
        popped = self.top.data
        self.top = self.top.next
        print(f"{popped} popped from stack")
        return popped

    def peek(self):
        if self.top is None:
            print("Stack is empty")
            return None
        return self.top.data

    def display(self):
        if self.top is None:
            print("Stack is empty")
        else:
            current = self.top
            print("Stack (top to bottom):", end=" ")
            elems = []
            while current:
                elems.append(current.data)
                current = current.next
            print(elems)

# Example usage:
stack = StackLL()
stack.push(10)
stack.push(20)
stack.push(30)
stack.display()  # Output: Stack (top to bottom): [30, 20, 10]
print("Top element:", stack.peek())  # Output: 30
print("Popped element:", stack.pop())  # Output: 30
stack.display()  # Output: Stack (top to bottom): [20, 10]
```

```python
# Implementing stack using deque

from collections import deque

class Stack:
    def __init__(self):
        self.stack = deque()

    def push(self, data):
        """Pushes an element onto the stack."""
        self.stack.append(data)

    def pop(self):
        """Removes and returns the top element of the stack."""
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.stack.pop()

    def peek(self):
        """Returns the top element without removing it."""
        if self.is_empty():
            print("Stack is empty")
            return None
        return self.stack[-1]

    def is_empty(self):
        """Checks if the stack is empty."""
        return len(self.stack) == 0

    def size(self):
        """Returns the size of the stack."""
        return len(self.stack)

    def display(self):
        """Displays the stack."""
        print("Stack (top to bottom):", list(self.stack)[::-1])

# Example usage
stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)
stack.display()  # Output: Stack (top to bottom): [30, 20, 10]
print("Top element:", stack.peek())  # Output: 30
print("Popped element:", stack.pop())  # Output: 30
stack.display()  # Output: Stack (top to bottom): [20, 10]
```
